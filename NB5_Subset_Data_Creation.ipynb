```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

np.random.seed(42)
```

```python
predictions = pd.read_csv('final_predictions.csv')
predictions[60:66]
```

```python
predictions_cleaned = predictions.dropna(subset=['pred_label'])

df_rest, df_sample = train_test_split(
    predictions_cleaned, # Use the cleaned DataFrame
    test_size=700,                 # ambil PAS 700 row
    stratify=predictions_cleaned['pred_label'],     # stratified by pred_label
    random_state=42                 # biar replikasi
)
```

```python
# Save to excel table (.xlsx)
df_sample[['rating', 'title', 'text', 'review_id']].to_excel('sample_predictions.xlsx', index=False)
```
